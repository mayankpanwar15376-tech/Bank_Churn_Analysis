# Step 5 - Feature Engineering
Create meaningful feature: e.g balance per products, age group, salary bracket and handle missing values.


```python
# Feature Engineering examples
df_fe= df.copy()

# Balance per product
df_fe['balance_per_product'] = df_fe['balance'] / (df_fe['products_number'].replace(0, np.nan)) 
df_fe['balance_per_product'].fillna(0, inplace=True)  # Replace NaN with 0 for customers with 0 products

# Salary to balance ratio
df_fe['salary_balance_ratio'] = df_fe['estimated_salary'] / (df_fe['balance'].replace(0, np.nan))
df_fe['salary_balance_ratio'].replace([np.inf, -np.inf], np.nan, inplace=True)  # Replace inf with NaN
df_fe['salary_balance_ratio'].fillna(df_fe['salary_balance_ratio'].median(), inplace=True)  # Replace NaN with median

# Age group
bins = [0, 25, 35, 45, 55, 65, np.inf]
labels = ['<25', '25-34', '35-44', '45-54', '55-64', '65+']
df_fe['age_group'] = pd.cut(df_fe['age'], bins=bins, labels=labels)

# Tenure group
df_fe['tenure_group'] = pd.cut(df_fe['tenure'], bins=[-1,0,2,5,10,100], labels=['0','1-2','3-5','6-10','10+'])

# Flag high balance
df_fe['high_balance'] = (df_fe['balance'] > df_fe['balance'].quantile(0.75)).astype(int)

# Quick checks
df_fe[['balance_per_product', 'salary_balance_ratio','age', 'age_group', 'tenure', 'tenure_group', 'high_balance']].head()


```

# Step 6 - Preprocessing - encoding & scaling

```python
# Define features and target
target = 'churn'
drop_cols = ['customer_id']
features = [c for c in df_fe.columns if c not in [target] + drop_cols]

numeric_features = ['credit_score', 'age', 'tenure', 'balance', 'products_number', 'estimated_salary',
                    'balance_per_product', 'salary_balance_ratio']
categorical_features = ['country', 'gender', 'credit_card', 'active_member', 'age_group', 'tenure_group', 'high_balance']

numeric_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

categorical_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])

preprocessor = ColumnTransformer([
    ('num', numeric_transformer, numeric_features),
    ('onehot', categorical_transformer, categorical_features)
])

print('Numeric features:', numeric_features)
print('Categorical features:', categorical_features)
```

# Step 7 - Train Test split

```python
x = df_fe[features]
y = df_fe[target]

X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42, stratify=y)
print('Train shape:', X_train.shape, 'Test shape:', X_test.shape)
print('Train churn distribution:', y_train.mean(), 'Test churn distribution:', y_test.mean())
```

# Step 8 - Train Multiple models with a pipeline and compare using cross-validation

```python
models = {
    'Logistic Regression': LogisticRegression(max_iter=1000, random_state=42),
    'Random Forest': RandomForestClassifier(n_estimators=200, random_state=42),
    'Gradient Boosting': GradientBoostingClassifier(n_estimators=200, random_state=42),
    'AdaBoost': AdaBoostClassifier(n_estimators=200, random_state=42),
    'SVC': SVC(probability=True, random_state=42)
}    

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
results = {}

for name, model in models.items():
    pipe = Pipeline(steps=[('preprocessor', preprocessor),
                            ('classifier', model)])
    # build a preprocessor that only uses columns present in X_train (avoids KeyError)
    num_present = [c for c in numeric_features if c in X_train.columns]
    cat_present = [c for c in categorical_features if c in X_train.columns]

    local_preprocessor = ColumnTransformer(
        transformers=[
            ('num', numeric_transformer, num_present),
            ('onehot', categorical_transformer, cat_present)
        ],
        remainder='drop'
    )

    pipe = Pipeline(steps=[('preprocessor', local_preprocessor),
                           ('classifier', model)])
    scores = cross_val_score(pipe, X_train, y_train, cv=cv, scoring='roc_auc', n_jobs=-1)
    results[name] = scores
    print(f"{name} AUC: Mean={scores.mean():.4f}, Std={scores.std():.4f}")
```

```python
# Boxplot of CV AUC scores
plt.figure(figsize=(8, 5))
sns.boxplot(data=[results[m] for m in list(results.keys())])
plt.xticks(ticks=range(len(results)), labels=list(results.keys()))
plt.ylabel('ROC AUC')
plt.title('Model Comparison (cross-validated ROC AUC)')
plt.show()
```

# Step 9 - Fit best model on full train set and evaluate on test set
Select best model by mean CV AUC above and evaluate standard metrics on the test set

```python
# Choose the best model (automatic pick by mean AUC)
best_name = max(results.keys(), key=lambda k: results[k].mean())
best_name, results[best_name].mean()
```

```python
best_model = models[best_name]

safe_numeric = [c for c in numeric_features if c in X_train.columns]
safe_categorical = [c for c in categorical_features if c in X_train.columns]

safe_preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, safe_numeric),
        ('onehot', categorical_transformer, safe_categorical)
    ],
    remainder='drop'
)

best_pipeline = Pipeline(steps=[('preprocessor', safe_preprocessor),
                                ('classifier', best_model)])
best_pipeline.fit(X_train, y_train)

# Predictions
y_pred = best_pipeline.predict(X_test)
y_proba = best_pipeline.predict_proba(X_test)[:, 1]

# Metrics
acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred)
rec = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
roc = roc_auc_score(y_test, y_proba)

print(f"Test Accuracy: {acc:.4f}")
print(f"Test Precision: {prec:.4f}")
print(f"Test Recall: {rec:.4f}")
print(f"Test F1-Score: {f1:.4f}")
print(f"Test ROC AUC: {roc:.4f}")

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title(f'Confusion Matrix - {best_name}')    
plt.show()
```
